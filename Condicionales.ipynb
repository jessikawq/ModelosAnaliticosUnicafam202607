Crea un programa que verifique si una persona es mayor de edad. Una persona es considerada mayor de edad si tiene 18 años o más.

```python
print("Verifiquemos si eres mayor de edad")
edad = int(input("Ingresa tu edad "))
if edad >= 18:
  print("Eres mayor de edad")
else:
  print("No eres mayor de edad")
```

Escribe un programa que verifique si un número ingresado por el usuario es par o impar.

```python
print("Verifiquemos si el número es par o impar")
numero = int(input("Ingresa el número "))
a = numero % 2
if a == 0:
  print("El número es par")
else:
  print("El número es impar")
```

Escribe un programa que clasifique la edad en diferentes categorías: Niño (0-12 años), Adolescente (13-17 años), Adulto (18-64 años) y Anciano (65 años o más).

```python
print("Clasifiquemos tu categoría de acuerdo a tu edad")
edad = int(input("Ingresa tu edad "))
if edad >=0 and edad <=12:
  print("Eres un niño")
elif edad >=13 and edad <=17:
  print("Eres un adolescente")
elif edad >=18 and edad <=64:
  print("Eres un adulto")
else:
  print("Eres un anciano")
```

Crea un programa que lea dos números y determine cuál es mayor, cuál es menor o si ambos son iguales.

```python
print("Clasifiquemos los numeros")
num1 = int(input("Ingresa primer numero "))
num2 = int(input("Ingresa segundo numero "))

if num1 > num2:
  print("El número ",num1, " es mayor que el numero", num2)
elif(num1 == num2):
  print("El número ",num1, " es igual al número", num2)
else:
  print("El número ",num2, " es mayor que el número", num1)
```

Crea un programa que determine si un número ingresado por el usuario está en el rango de 1 a 100 (inclusive) y si es divisible por 5.

```python
print("El numero esta en el rango ?")
numero = int(input("Ingresa el número "))
a = numero % 5
if numero <= 100 and a == 0:
  print("Si esta en el rango y es divisible por 5")
else:
  print("fuera de rango o no es divisible por 5")
```
