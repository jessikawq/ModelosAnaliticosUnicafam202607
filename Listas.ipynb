##LISTAS

```python
#Una lista vacía
mi_lista= []
#Lista con varios elementos
mi_lista=[1,2,3,"4",5.0,True]
print(mi_lista[2])
```

```python
#Cambios en la lista
mi_lista[0]="jessika"
print(mi_lista)
```

```python
#Agregar elementos a una lista
mi_lista.append("Rodriguez")
print(mi_lista)
```

```python
#Agregar elementos a una posicion especifica
mi_lista.insert(3,"Bogota")
print(mi_lista)
```

```python
#Agregar varios elementos al final de la lista
mi_lista.extend([7,False])
print(mi_lista)
```

```python
# Quitar elementos
# Eliminar objetos especificos
mi_lista.remove(True)
print(mi_lista)
```

```python
print("ACTUALIZACION DE DATOS")
valor_buscar=input("por favor ingrese el nombre para actualizar")
valor_reemplazar=input("ingrese valor a remplazar:")
mi_lista[mi_lista.index(valor_buscar)]=valor_reemplazar
print("cambio ejecutado. Imprimientdo resultado")
print(mi_lista)
```
