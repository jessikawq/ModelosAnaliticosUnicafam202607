Taller Actualización de datos

```python
#Diccionario con datos básicos
Estudiante = {
    "Nombres":"Jessika Mharia" ,
    "Apeliilos":"Rodriguez",
    "Ciudad":"Bogotá",
    "Hijos":"No",
    "Carrera":"Programación"
}
```

```python
# Contruccion del menu del usuario
print("---- Bienvenido a tu programa de actualización de datos----")
print("Para comenzar elige la opción deseada")
print("1: Agregar un datos nuevo")
print("2: Actualizar un dato")
print("3: Eliminar un dato con del")
print("4: Eliminar un dato con pop")
print("5: Consultar un dato")
print("6: Imprimir diccionario completo")
seleccion =int(input("Selecciona la opcion deseada:"))

```

```python
# Agregar un dato nuevo
if seleccion == 1:
  clave = input("Ingrese la clave deseada:")
  valor = input("Ingrese el valor deseado:")

  # Agregar elemento nuevo
  Estudiante[clave] = valor
  print("El valor se agrego exitosamente")
  print(Estudiante)
# Actualizar
elif seleccion == 2:
  clave = input("Ingrese la clave a actualizar:")
  Estudiante[clave] = input("Ingrese valor nuevo")
  Estudiante[clave] = ("El dato se actualizado con exito")
  print(Estudiante)

# Eliminar
elif seleccion == 3:
  clave = input("Ingrese la clave a eliminar con del")
  del Estudiante[clave]
  print("El dato fue eliminado exitosamente!!")
  print(Estudiante)

elif seleccion == 4:
  clave = input("Ingrese la clave a eliminar con pop")
  Estudiante.pop(clave)
  print("El dato fue eliminado exitosamente")
  print(Estudiante)

#Consultar dato
elif seleccion == 5:
  clave = input("Ingrese clave a consultar")
  valor = Estudiante.get(clave,"No disponible")
  print(f"{clave}:{valor}")

#Imprimir Diccionario completo
elif seleccion == 6:
  print("Perfil completo")
  print(Estudiante)

else:
  print("Opción invalida")


```
